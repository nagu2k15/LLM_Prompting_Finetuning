```python
# remove stopwords
# common word
# tokenization
# remove @virginamerica #va - virginamerica, va
# stemming/lemmatzation
# remove special characters
import re
def remove_special_charcters(text):
    pattern = r'[^a-zA-z0-9\s]' # readlines,^ - NOT starting with a-zA-Z0-9
    text = re.sub(pattern, '', text) # look for the pattern in the text, substitute with ''
    return text
# print(remove_special_charcters(text='@VirginAmerica SFO-PDX schedule is still MIA.'))
df_subset['text'] = df_subset['text'].apply(remove_special_charcters)
# incase of special character removals
# df_subset['text'] = df_subset['text'].apply(lambda x: str(x).replace('!',''))
df_subset.head()
# stemming is very crude, social media

## hmmmmm.......... lol.......

## email emoticons, new articles
```

```python
from nltk.stem.porter import PorterStemmer # importing stemmer
# instaiting the object
ps = PorterStemmer()
# sentences = ["iam doing great",
#              "i like my students", "students are really good"]
# initializing an empty list
text_lemma = []
for sentence in df_subset['text']:
  # for loop, will take one row at a time
  text_lemma.append(ps.stem(sentence))
  # porter stemmer for the sentence/row and result of the porter stemmer is appended to the list
  # creating a new column and storing the text_lemma in to that column
df_subset['text_lemma'] = text_lemma
```

```python
from sklearn.preprocessing import LabelEncoder
le= LabelEncoder()
df_subset['sentiment_label'] = le.fit_transform(df_subset['airline_sentiment'])
df_subset.head()
# x data is a number - ydata - number
# 2784628*75982+1 = 1
# Positive, negative and neutral
# [A-Z sorting] - negative [0], neutral [1] and Positive [2]
```

```python
# Divide the data in to training and testing
# train a model - 80 rows of data out of 100
# test the model- 20 rows put aside are used for testing
from sklearn.model_selection import train_test_split #library
x_train,x_test,y_train,y_test = train_test_split(df_subset['text_lemma'],
                                                 df_subset['sentiment_label'],
                                                 test_size=0.2, #20% of data is for test data
                                                 random_state=0) # randomly, seed number = 0, use the same seed number, same rows of data in training and testing data
# random_state =0 , [1,3,5,2] - train , [4]- testing
# random_state =3 , [1,4,5,2] - train , [3]- testing
print(x_train.shape,x_test.shape,y_train.shape,y_test.shape)
# test data size seems to be same as train data. How the test data is only 20%?
```

```python
# stopwords, lower case, tokenization
# To convert unstructured text data to structured format [array of numbers]
# CountVectorization can be used
from sklearn.feature_extraction.text import CountVectorizer
# importing the library
cv = CountVectorizer(ngram_range=(1,1), stop_words="english") #n_gram_range = 1,2 means, consider unigram, and bigrams
# instatiating the class
# fit is run - extract the features [unique words in the entire data - training data]
# transform is happening, only mapping is done
x_train_with_stop = cv.fit_transform(x_train).toarray()
# fit_transform for train data, but only tranform for test data?
x_test_with_stop = cv.transform(x_test).toarray()
# print(x_train_without_stop.shape)
# print(x_test_without_stop.shape)
# n_grams: bigram, trigram
# fit is run - extract the features [unique words in the entire data - training data]
# bigrams - [fit is, is run, rin-extract, extract the]
#
# what does this 15105 in training data of X?
# features have been identified, with fit_transform,
# so now we eed only to do the count againt the features..
# age | gender --> income
```

```python
# supervised learning / unsupervised learning
# sentiment analysis - text data and sentiment
# x data [text] and y data [ sentiment]
# supervised learning
## regression - output is a continous number [profit, revenue, house price prediction]
## classification - [categories -limited number, classes]
## clustering - unsupervised - no ydata
## Naive bayes - f1, f2, f3, f4, f5 .. fn
## sentiment - positive, negative and neutral
## assign a score to each word depending on the importance of the word to the class
## Naive bayes - conditional probability
## p(a/b) = p(a intersection b)/p(b) - # p(mohan reaching office late/he woke late)
## p("excellent"/"positive") = 0.9 - training
## p("excellent"/"negative") = 0.02
## inference - "Movie is excellent" - p(s=positive/"excellent")
from sklearn.naive_bayes import MultinomialNB
# loaded the library for metric calculation
from sklearn import metrics
# instatiated the multinomial naive bayes
nb = MultinomialNB()
# x_train and y_train - learn the patterns in the data
nb.fit(x_train_with_stop,y_train) # training
# y = w1x1+w2x2+5
# yprediction =
# x_test  - transform/predict
# 99% of data for training - training data - accuracy - 100%
## 1% test data [does not know] - model- predicitons - 60%
```

```python
# predict attribute of nb trained model on x_test
y_pred_test = nb.predict(x_test_with_stop)
# srtoring the prediciton results
df_metrics = pd.DataFrame(y_pred_test, columns=['prediction'])
metrics.accuracy_score(y_test.values,y_pred_test)
# ytest values are actual and ypred are model predicted
#metrics.confusion_matrix(y_test.values,y_pred_test)
```
